# Filtering Queries with UDTFs

This notebook demonstrates how to filter data when querying UDTFs using WHERE clauses.

## Prerequisites

- UDTFs must be registered (see `registration.ipynb`)
- CDF credentials in `config.toml` file


## Step 0: Load Configuration


```python
import tomli

# Load credentials from TOML file
with open("config.toml", "rb") as f:
    config = tomli.load(f)

cognite_config = config["cognite"]
```

## Step 1: Equality Filters


```python
# Filter by single property - select all properties
query = f"""
SELECT * FROM smallboat_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => NULL,
    description => NULL
)
WHERE name = 'MyBoat'
LIMIT 10;
"""

result = spark.sql(query)
result.show(truncate=False)
```

```python
# Filter by space and external_id
query = f"""
SELECT * FROM vessel_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => NULL,
    description => NULL
)
WHERE space = 'sailboat' AND external_id = 'vessel-123'
LIMIT 10;
"""

result = spark.sql(query)
result.show(truncate=False)
```

## Step 2: NULL Handling


```python
# Filter out NULL values
query = f"""
SELECT * FROM smallboat_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => NULL,
    description => NULL
)
WHERE description IS NOT NULL
LIMIT 10;
"""

result = spark.sql(query)
result.show(truncate=False)
```
