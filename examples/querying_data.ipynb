# Querying Data with UDTFs

This notebook demonstrates various ways to query data using UDTFs in Spark.

## Prerequisites

- UDTFs must be registered (see `registration.ipynb`)
- CDF credentials in `config.toml` file


## Step 1: Load Configuration


```python
import tomli

# Load credentials from TOML file
with open("config.toml", "rb") as f:
    config = tomli.load(f)

cognite_config = config["cognite"]
print("✓ Configuration loaded")
```

## Step 2: Query Single UDTF


```python
# Basic query with LIMIT
query = f"""
SELECT * FROM smallboat_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => NULL,
    description => NULL
) LIMIT 10;
"""

result = spark.sql(query)
result.show(truncate=False)
```

## Step 3: Select All Properties vs Specific Properties


```python
# Select all properties (SELECT *)
query_all = f"""
SELECT * FROM smallboat_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => NULL,
    description => NULL
) LIMIT 10;
"""

result_all = spark.sql(query_all)
print("All properties:")
result_all.show(truncate=False)
```

```python
# Select specific properties (subset of columns)
query_subset = f"""
SELECT 
    external_id,
    name,
    space
FROM smallboat_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => NULL,
    description => NULL
) LIMIT 10;
"""

result_subset = spark.sql(query_subset)
print("Specific properties only:")
result_subset.show(truncate=False)
```

## Step 4: Named Parameters


```python
# Named parameters (recommended - clearer and more maintainable)
query_named = f"""
SELECT * FROM smallboat_udtf(
    client_id => '{cognite_config["client_id"]}',
    client_secret => '{cognite_config["client_secret"]}',
    tenant_id => '{cognite_config["tenant_id"]}',
    cdf_cluster => '{cognite_config["cdf_cluster"]}',
    project => '{cognite_config["project"]}',
    name => 'MyBoat',
    description => NULL
) LIMIT 10;
"""

result_named = spark.sql(query_named)
print("Named parameters:")
result_named.show(truncate=False)
```
